```python
import pandas as pd
import sqlite3

conn = sqlite3.connect(r"C:\Users\Francisco Neto\Desktop\brazilian-ecommerce\database\olist.db")

df = pd.read_sql("SELECT * FROM vendedores", conn)

# Transformação título nas cidades
df["seller_city"] = df["seller_city"].str.title()

# Sobrescreve a tabela com os dados tratados
df.to_sql("vendedores", conn, if_exists="replace", index=False)

print(df["seller_city"].head(10))
conn.close()
```
